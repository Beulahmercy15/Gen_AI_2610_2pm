### **Python If,Elif, Else(Decision statement)**
### **If ,elif, else rules**
* if (Only once per code block(Mandtory))
* elif (as many time as per code block(optional))
* else (Only once per code black(optional))

*Syntax*
* if \<condition>:
* elif \<condition>:
* else:

### **if \<condition>:**

```python
a = 101
if a>100:
    print(a,"is bigger then 100")
print("This line alwas work")
```

### **else:**

```python
a = 10
if a>100:
    print(a,"is bigger then 100")
else:
    print(a,"is smaller then 100" )
    
print("This line alwas work")
```

### **if \<condition>:**
### **elif \<condition>:**
### **else:**

```python
a = 10
if a>100:
    print(a,"is bigger then 100")
elif a == 100:
    print(a,"is equal to 100")
else:
    print(a,"is smaller then 100" )
    
print("This line alwas work")
```

**elif(as many time as per code block(optional))**


```python
a = 30
if a > 50:
    print("a is bigger")
elif a == 20:
    print("a is 20")
elif a == 30:
    print("a is 30")
elif a == 40:
    print("a is 40")
else:
    print("a is smaller")
    
print("Code Over")
```

### **Nested If and else**

```python
a = 100
if a>10:
    print("a is bigger")
    if a>=100:
        print("a is BIGGEST")
    else:
        print("a btwn 10-99")
elif a<10:
    print("a is smaller")
    if a<0:
        print("a is negative")
    else:
        print("a btwn 0-9")
print("code over")
```

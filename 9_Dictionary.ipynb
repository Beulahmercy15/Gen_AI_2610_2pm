### **Dictionary Functions**

```python
mydict ={
"Name":"Beulah Mercy",
"City":"Chennai",
"Course":"Python",
500 : 50.55,
}
print(mydict,type(mydict))
print(mydict["Course"])
mydict["Age"]=40
print(mydict)
```

*keys{} Returns a list containing the dictionary's keys*

```python
mydict ={
"Name":"Beulah Mercy",
"City":"Chennai",
"Course":"Python",
500 : 50.55,
}
print(mydict)
print(list(mydict.keys()))
print("-----------")
for each in mydict.keys():
    print(each)
```

*values{} Returns a list of all the values in the dictionary*

```python
mydict ={
"Name":"Beulah Mercy",
"City":"Chennai",
"Course":"Python",
500 : 50.55,
}
print(mydict)
print(list(mydict.values()))
print()
for each in mydict.values():
    print(each)
```

*items{} Returns a list containing a tuple for each key value pair*

```python
mydict ={
"Name":"Beulah Mercy",
"City":"Chennai",
"Course":"Python",
500 : 50.55,
}
print(mydict)
print(list(mydict.items()))
print()
for each in mydict.items():
    print(each)
```

```python
mydict ={
"Name":"Beulah Mercy",
"City":"Chennai",
"Course":"Python",
500 : 50.55,
}
print()
for each in mydict.items():
    print(f"{each[0]}\t{each[1]}")
```

*pop(key) Removes the element with the specified key*

```python
mydict ={
"Name":"Beulah Mercy",
"City":"Chennai",
"Course":"Python",
500 : 50.55,
}
print(mydict)
mydict.pop("Course")
mydict.pop("Name")
print(mydict)
```

*popitem{} Removes the last inserted key-value pair*

```python
mydict ={
"Name":"Beulah Mercy",
"City":"Chennai",
"Course":"Python",
500 : 50.55,
}
print(mydict)
mydict.popitem()
mydict.popitem()
print(mydict)
```

*clear() Removes all the elements from the dictionary*

```python
mydict ={
"Name":"Beulah Mercy",
"City":"Chennai",
"Course":"Python",
500 : 50.55,
}
print(mydict)
mydict.clear()
print(mydict)
```

*update{} Updates the dictionary with the specified key-value pairs*

```python
mydict ={
"Name":"Beulah Mercy",
"City":"Chennai",
"Course":"Python",
500 : 50.55,
}
print(mydict)
newdict = {"Name":"Honey","City":"Mumbai","Course":"Java","age":35}
mydict.update(newdict)
print(mydict)
```

*get{} Returns the value of the specified key*

```python
mydict ={
"Name":"Beulah Mercy",
"City":"Chennai",
"Course":"Python",
500 : 50.55,
}
print(mydict)
print(mydict[500])
print(mydict[600])
print("Hello Python")
```

```python
mydict ={
"Name":"Beulah Mercy",
"City":"Chennai",
"Course":"Python",
500 : 50.55,
}
print(mydict)
print(mydict.get(500))
print(mydict.get(600))
print("Hello Python")
```

*copy{} Returns a copy of the dictionary*

```python
x = {"A":100,"B":200,"C":300}
y = {"A":100,"B":200,"C":300}
x["D"]=400
print(x,type(x),id(x))
print(y,type(y),id(y))
```

```python
x = {"A":100,"B":200,"C":300}
y = x.copy()
x["D"]=400
y["E"]=500
print(x,type(x),id(x))
print(y,type(y),id(y))
```

### **Nested Dictionary**

```python
mydict = {
"Name": "Beulah Mercy",
"City":["Chennai","Kovai","Kanchi","Mumbai"],
"Year":(2000,2010,2020)
}
print(mydict,type(mydict))
print(mydict.keys())
print(mydict.get("City"))
print(mydict.get("City")[0])
print(mydict.get("City")[2])
print("---------------")
print(mydict.get("Year"))
print(mydict.get("Year")[1])
print("---------------")
print(mydict.get("City")[3][3:])
```

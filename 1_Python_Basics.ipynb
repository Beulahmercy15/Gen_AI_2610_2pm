### **Trio : value, id, type**

* value - eg a = 10; a - variable, = - assign, 10 - value
* id    - eg a = 10; 10 value stored in 'a' variable; 'a' address is named as id
* type  - eg a = 10; the 10 is integer (int)*

```python
a = "100"
b = 100
c = 15.7
d = True
print(a)
print(b)
print(c)
print(d)
print(type(a))
print(id(a))
print(type(b))
print(id(b))
print(type(c))
print(id(c))
print(type(d))
print(id(d))
```

### **Python Type Casting**
* one data type can be converted to another data type.

```python
# int()

a = 1000
a1 = float(a)
a2 = complex(a)
a3 = str(a)
a4 = bool(a) # True <class 'bool'>

# float()

b = 10.11
b1 = int(b)
b2 = str(b)
b3 = bool(b) # True <class 'bool'>

# str("Character")

c = "Mercy" # c1 = int(c) # ValueError: invalid literal for int() with base 10: 'hello'
            # c2 = float(c)  # ValueError: could not convert string to float: 'hello'
c3 = bool(c) # True <class 'bool'>

# str("Numeric")

d = "600"
d1 = int(d)
d2 = float(d)
d3 = bool(d) # True <class 'bool'>

# bool false statement

e = 0
e1 = bool(e) # False
e2 = "" # False
e3 = bool(e2)
print(e3,type(e3))
```

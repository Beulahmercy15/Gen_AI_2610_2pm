### **Python Operators**

```python
# Arithmetic operators -  +, -, *, / ,** ,//,%

print(10 + 3)  # 13
print(10-3)    #  7
print(10 * 3)  # 30
print(10 / 3)  # 3.3333...
print(10 ** 3) # pow => 10 * 10* 10 = 1000
print(10 // 3) # Q => 3
print(10 % 3)  # R => 1
```

```python
# Assignment operators -  =, +=, -=, /=, *= , **=, //=, %=

a = 10
print(a)
a += 5 # a = a + 5 = 15
a *= 2 # a = 15 * 2 = 30

b = 0
b += 50

print(b)
```

```python
# Comparison operators - (True or False)
# ==, !=, >, <, <=, >=

print(10 == 10)# True
print(10 != 10)# Flase
print(10 > 10) # Flase
print(10 >= 10)# True
print(10 < 10) # False
print(10 <= 10)# True
```

```python
# Logical operators => (True or False)
# and, or, not
# and(multiple) , or(add) , not(inverse)
""" and :                OR :                 NOT : 
      T and T = T           T or T = T            NOT T = F
      T and F = F           T or F = T            NOT F = T
      F and T = F           F or T = T
      F and F = F           F or F =                        T"""

print("-------AND--------")

print(10 == 10 and 6 > 4) # T AND T = T
print(10 != 10 and 6 > 4) # F AND T = F
print(10 == 10 and 6 < 4) # T AND F = F
print(10 != 10 and 6 < 4) # F AND F = F

print("-------or--------")

print(10 == 10 or 6 > 4) # T OR T = T
print(10 != 10 or 6 > 4) # F OR T = T
print(10 == 10 or 6 < 4) # T OR F = T
print(10 != 10 or 6 < 4) # F OR F = F

print("-------NOT--------")

print(not 10 == 10) # F
print(not 10 != 10) # T
```

```python
# while loop

a = 1
while a<=5:
    b = 1
    while b<=5:
        print("*",end='')
        b += 1
    a += 1
    print()
```

```python
row = int(input("Enter the number:"))
col = int(input("Enter thhe number:"))
symbol = input("Enter the symbol")
a=1
while a<=row:
    b=1
    while b<=col:
        print(symbol,end="")
        b+=1
    a +=1
    print()
```

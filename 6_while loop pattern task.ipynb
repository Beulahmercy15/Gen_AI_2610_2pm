```python
a = 1
while a<=5:
    b=1
    while b<=a:
        print("#",end='')
        b+=1
    a+=1
    print()
```

```python
a = 1
while a<=5:
    b=2
    while b<=a+1:
        print("#",end='')
        b+=1
    a+=1
    print()
```

### **print the blow pattern**

```python
for x in range(5):
    for y in range(5-x):
        print("&",end="")
    print()
```

```python
a=5
while a>0:
    b=0
    while b<a:
        print("&",end="")
        b+=1
    a-=1
    print()
```

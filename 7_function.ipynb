### **User Defind function**
*Syntax*

**def \<func_name>():**

```python
def happy(): # func creation
    print("I\'am Happy")
    print("Code reusability")
    
happy() # func calling
```

```python
def happy():
    print("I\'am Happy")
    print("Code reusability")
def sad():
    print("I\'am Sad")
    print("Code reusability")
sad()
print("----------")
happy()
print('-----------')
sad()
```

### **function with Parameters**

```python
def add():
    x = 10
    y = 20
    print(x+y)
add()
add()
```

```python
def add(x,y):# x,y => Parameters
    print("X =",x)
    print("Y =",y)
    print("The sum of :",x+y)
add(40,20) # 40,20 => arguments
print()


add(100,300)
print()
add(500,0)
```

```python
def myfunc(name,age):
    print("My Name :",name)
    print("My age :",age)
myfunc("Manju",40)
```

### **function with default Parameters**

```python
def add(x=0,y=0):
    print("X =",x)
    print("Y =",y)
    print("The sum of :",x+y)
add()
print()
add(500,600)
print()
add(y=60)
```

```python
def myfunc(name= "No Name",age = "No Age"):
    print("My Name :",name)
    print("My age :",age)
myfunc("Manju",40)
myfunc("Honey")
```

### **Function with return keyword**

```python
def myfunc(x=0,y=0):
    print("X =",x)
    print("Y =",y)
    print("The sum of :",x+y)
ans = myfunc(50,70)
print(ans)
```

```python
ans = 50+70
print(ans)
```

```python
def myfunc(x=0,y=0):
    print("X =",x)
    print("Y =",y)
    return x+y
ans = myfunc(50,70)
print(ans)
ans2 = myfunc(200,300)
print(ans2)
print(myfunc(30,20))
```

```python
def myadd(x=0,y=0):
    return x+y
def mysub(x=0,y=0):
    return x-y
#            110           50
ans = mysub(myadd(50,60),myadd(30,20))
print(ans)
```

### **Arbitrary Arguments**
*Types of arguments*

* \*args - argument stores in tuple
* \*\* kwargs - keyworded argument stores in dictionary

```python
def myfunc(*x): # *args - argument stores in tuple
    print(x,type(x))
myfunc()
myfunc(10)
myfunc("hai")
myfunc(50.55)

myfunc(10,30.55,"hai")
```

```python
def myfunc(**x): # **kwargs - keyworded argument stores in dictionary
    print(x,type(x))
myfunc(a = 10,b = 100,name = "Jeevan")
```

### **Function overloading**

```python
def myfunc(x):
    return x
def myfunc(x,y):
    return x+y
def myfunc(x,y,z):
    return x+y+z
print(myfunc(100,40,50))
```

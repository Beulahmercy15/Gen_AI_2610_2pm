```python
# Interview check palindrome
# 1 Find weather the String is check palindrome
# 1. Input: racecar
# Output: True
# 2. Input: MalaYa lAm
# Output: True
# 3 Input: Sir, I demand, I am a maid named Iris
# Output: True
# 4 Input : Red roses run no risk. Sir, on Nurse's Order.
# Output: True
```

*common*

```python
mystr = 'racecar'
revstr = mystr[::-1]
print(mystr)
print(revstr)
if mystr == revstr:
    print(True)
else:
    print(False)
```

*if class*

```python
mystr = 'MalaYa lAm'
newstr = mystr.lower().replace(' ','')
revstr = newstr[::-1]
print(mystr)
print(newstr)
print(revstr)
if newstr == revstr:
    print(True)
else:
    print(False)

```

```python
mystr = 'Sir, I demand, I am a maid named Iris'
newstr = mystr.lower().replace(' ','').replace(',','')
revstr = newstr[::-1]
print(mystr)
print(newstr)
print(revstr)
if newstr == revstr:
    print(True)
else:
    print(False)
```

```python
mystr = "Red roses run no risk. Sir, on Nurse's Order."
newstr = ''
for each in mystr:
    if each.isalpha():
        newstr= newstr+each.lower()
if newstr == newstr[::-1]:
    print(True)
else:
    print(False)
```

*function*

```python
def Check_palindrome(mystr):
    newstr = ''
    for each in mystr:
        if each.isalpha():
            newstr= newstr+each.lower()
    if newstr == newstr[::-1]:
        return True
    else:
        return False
ans = Check_palindrome("Red roses run no risk. Sir, on Nurse's Order.")
print(ans)
print(Check_palindrome("racecar"))
print(Check_palindrome("MalaYa lAm"))
print(Check_palindrome("Sir, I demand, I am a maid named Iris"))
```

### **Character frequency**

*common*

```python
# input = "Hello WOR Ld"
# output={'h':1, 'e':1, 'l':3, 'o':2, 'w':1, 'r':1, 'd':1}
mystr = "Hello WOR Ld"
newstr =''
for each in mystr:
    if each.isalpha():
        newstr = newstr+each.lower()
mydict = {}
for each_1 in newstr:
    if each_1 not in mydict.keys():
        mydict[each_1]=1
else:
    mydict[each_1]+=1
print(mydict)
```

*function*

```python
def Character_frequency(mystr):
    newstr =''
    mydict = {}
    for each in mystr:
        if each.isalnum():
            newstr = newstr+each.lower()
    for each_1 in newstr:
        if each_1 not in mydict.keys():
            mydict[each_1]=1
        else:
            mydict[each_1]+=1
    return mydict
print(Character_frequency("Hello world"))
print(Character_frequency("ManjuNathan"))
print(Character_frequency("9840243168"))
```

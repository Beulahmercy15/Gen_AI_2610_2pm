### **WHILE LOOP**
*Syntax*

**while <\condition>:**

```python
a = 1                 # Initilization/Start
while a<=10:          # Condition/End
    print(a,end = " ")
    a+=1              #Incrementation/Jump
    
print("\nLoop over")
```

**Reverse**

```python
a = 10
while a>=1:
    print(a,end=' ')
    a-=1
    
print("\nLoop over")

```

### **Keywords of while:**
* break
* continue

### **break - stop the loop**


```python
a= 1
while a>0:
    if a == 20:    #break condition
        print(a)
        break
    print(a, end=' ')
    a+=1

print("\nLoop Over")
```

### **continue - go back to the top of my loop**

```python
a = 1
while a<=20:
    if a % 3 == 0:
        a += 1
        continue
    print(a,end= " ")
    a +=1
print("\nLoop over")
```

```python
a = 1
while a<=20:
    if a % 3 == 0:
        print(a,end= " ")
        a += 1
        continue
    
    a +=1
print("\nLoop over")
```

### **Nested While Loop**


```python
"""
12345
12345
12345
12345
12345
"""
a = 1
while a<=5:
    b = 1
    while b<=5:
        print(b,end='')
        b += 1
    a += 1
    print()
```

user input

*row =*

*column =*

```python
row = int(input("Enter the row num :"))
col = int(input("Enter the col num :"))
a = 1
while a<= row:
    b = 1
    while b<= col:
        print(b,end = '')
        b += 1
    a += 1
    print()
```

### **For loop**
* For loop is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, or a string).

*Syntax*

**for \<variable_name> in \<Str\List\Tuple\Dict\Set>:**

```python
for x in "Hello world":  # Fetches Letter by letter
    print(x,end='')
print("\nLoop Over")
```

```python
for y in [10,15.55,True,"Mercy","pappu"]:
    print(y)
```

```python
mylist =[10,15.55,True,"Mercy","pappu"] # Fetches Item by Item
for each in mylist:
    print(each,end=', ')
```

### **range() – is a python standard function used for numbers**

*end before*

```python
for x in range(5+1): # end before
    print(x,end=' ')
```

*start with end before*

```python
for x in range(50,100+1): # start with end before
    print(x,end=' ')
```

*jump*

```python
for x in range(0,101,10): # jump
    print(x,end=' ')
```

```python
for x in range(100,-1,-10):
    print(x,end=" ")
```

### **for with else**


* If Loop completes successfully, else will worke
* If loop is unsuccessfully , else will not worke

```python
for x in "Hello world":
    print(x,end='')
else:
    print("\nThe above loop complieted successfully")
```

```python
for x in "Hello world":
    if x == ' ':
        break
    print(x,end='')
else:
    print("\nThe above loop complieted successfully")
```

### **find if a number is divisible by 5 in the list**

```python
mylist = [2,6,9,10,15,11,16,97]
for each in mylist:
    if each % 5 == 0:
        print(each,"Divisible By 5")
        break
else:
    print("Number divisible by 5 not found")
```

### **Nested for Loop**


```python
"""
12345
12345
12345
12345
12345
"""
for x in range(5):
    for y in range(1,6):
        print(y,end='')
    print()
```

```python
for x in ["Yellow","Green"]:
    print("-------------------------")
    for y in ["Cat","Dog","Car","House"]:
        print(x,y)
    print("-------------------------")
print("Loop Over")
```
